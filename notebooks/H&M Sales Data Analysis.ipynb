# Customer Intelligence Through Fashion: An H&M Data Analytics Case Study

## Project Overview

### Executive Summary

This project analyzes 2+ years of H&M transaction data to uncover customer purchasing patterns, develop customer segments, and build actionable recommendations for inventory management and marketing strategies. Using 31+ million transactions from 1.37 million customers across 106,000 products, this analysis demonstrates end-to-end data analytics capabilities applicable to retail and e-commerce environments.

### Business Context

H&M, one of the world's largest fashion retailers, faces the challenge of personalizing the shopping experience for millions of customers across thousands of products. With customers potentially overwhelmed by choice, understanding purchase patterns and predicting future behavior becomes critical for:

- Reducing decision fatigue
- Increasing conversion rates
- Optimizing inventory management
- Personalizing marketing campaigns

### Target Audience

This portfolio project is designed to demonstrate analytics capabilities to:

- Hiring managers seeking junior/mid-level data analysts
- Retail and e-commerce companies
- Teams focused on customer analytics and business intelligence

### Data Dictionary

#### articles.csv

- article_id: Unique product identifier
- product_name: Product description
- product_type_name: Category (e.g., Trouser, T-shirt)
- product_group_name: Higher-level grouping
- colour_group_name: Primary color classification
- department_name: Target department (e.g., Menswear, Womenswear)
- index_name: Store section
- section_name: Product section
- garment_group_name: Garment category
- detail_desc: Detailed product description

#### customers.csv

- customer_id: Unique customer identifier (hashed)
- age: Customer age (nullable)
- postal_code: Location identifier (hashed)
- FN: Fashion news subscription status (1/0)
- Active: Activity status (1/0)
- club_member_status: Membership tier

#### transactions_train.csv

- t_dat: Transaction date (2018-09-20 to 2020-09-22)
- customer_id: Customer identifier
- article_id: Product identifier
- price: Purchase price
- sales_channel_id: Channel (1=store, 2=online)
